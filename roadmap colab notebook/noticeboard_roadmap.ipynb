# Production Roadmap: Multi-Instance Noticeboard App

## Architecture Overview
- **Supabase** as the core database (PostgreSQL + auth + storage)
- **FastAPI** backend interacting with Supabase via Python client
- **Discord Bot** wrapper enabling users to update noticeboard data via Discord commands
- Multi-instance architecture to allow multiple servers/communities to have isolated noticeboards

---

## Phase 1 — Foundation
### 1. Supabase Setup
- Create project and configure Postgres database
- Define schema:
  - `instances` (id, name, created_at)
  - `notices` (id, instance_id, author, content, timestamp)
  - `discord_instances` (instance_id, guild_id)
- Configure row-level security
- Create service key and anon key

### 2. FastAPI Backend
- Initialize FastAPI project structure
- Implement Supabase client integration
- Core endpoints:
  - POST /instances
  - GET /instances/{id}/notices
  - POST /instances/{id}/notice
- Input validation with Pydantic
- Testing setup

---

## Phase 2 — Discord Bot Integration
### 3. Discord Bot Foundation
- Create bot with discord.py or nextcord
- Bot authentication via token
- Command handler system

### 4. Bot → FastAPI Integration
- Implement API calls for:
  - Creating new notices
  - Fetching notices for display in Discord
- Commands:
  - `/create_notice`
  - `/list_notices`
  - `/register_instance`

### 5. Multi-Instance Handling
- Detect guild ID
- Map guilds to Supabase instances
- Provide server admins a setup command

---

## Phase 3 — Deployment Pipeline
### 6. FastAPI Deployment
- Deploy on Fly.io or Railway
- Set environment variables securely
- Configure CORS for future frontend

### 7. Discord Bot Deployment
- Deploy via Docker container
- Auto-restart setup
- Logging and monitoring

---

## Phase 4 — Scaling & Improvements
### 8. Feature Enhancements
- Notice editing
- Image and file attachments
- Permissions per role

### 9. Performance & Observability
- Add caching layer (Redis optional)
- Add Sentry logging
- Improve Supabase RLS policies

---

## Phase 5 — Production Rollout
- End-to-end testing across multiple Discord servers
- Load testing FastAPI
- Documentation and admin guides
